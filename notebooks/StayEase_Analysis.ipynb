## Data Quality Check — Missing Values

## Data Quality Check — Duplicate Rows

Duplicate records can distort totals, averages, and other analysis.

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

CLEANING

```python
print("Missing CustomerName before cleaning:",
      df["CustomerName"].isnull().sum())
```

```python
df["CustomerName"] = df["CustomerName"].fillna("Unknown")
```

```python
print("Missing CustomerName after cleaning:",
      df["CustomerName"].isnull().sum())
```

```python
rating_median = df["Rating"].median()

print("Median Rating:", rating_median)
```

```python
df["Rating"] = df["Rating"].fillna(rating_median)
```

```python
print("Missing Rating after cleaning:",
      df["Rating"].isnull().sum())
```

```python
duplicate_count = df.duplicated().sum()

print("Duplicate rows before cleaning:", duplicate_count)
```

```python
df = df.drop_duplicates()
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print()
print("Missing values:")
print(df.isnull().sum())
print()
print("Duplicate rows:", df.duplicated().sum())
```

```python
df["City"].value_counts()
```

```python
print("Unique city values before cleaning:")
print(df["City"].unique())
```

```python
df["City"] = df["City"].str.title()
```

```python
print("Unique cities after cleaning:")
print(df["City"].unique())
```

```python
print("Number of unique cities:", df["City"].nunique())
```
