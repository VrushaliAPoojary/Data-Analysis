## 4.2 Correlation Analysis

Correlation measures the strength and direction of the linear relationship
between numerical variables.

```python
correlation_data = df[
    ["Sales", "Profit", "Quantity", "DeliveryDays"]
]

correlation_matrix = correlation_data.corr()

correlation_matrix
```

```python
corr_pairs = correlation_matrix.where(
    np.triu(np.ones(correlation_matrix.shape), k=1).astype(bool)
)

corr_pairs
```

```python
strongest_pair = (
    corr_pairs.stack()
    .abs()
    .sort_values(ascending=False)
)

strongest_pair
```

## 4.5 Total Sales by Region

We calculate the total sales generated by each region.

```python
sales_by_region = df.groupby("Region")["Sales"].sum()

sales_by_region
```

```python
sales_by_region_sorted = sales_by_region.sort_values(
    ascending=False
)

sales_by_region_sorted
```

```python
top_region = sales_by_region_sorted.index[0]
top_region_sales = sales_by_region_sorted.iloc[0]

print("Top region:", top_region)
print("Total sales:", top_region_sales)
```

## 4.6 Average Profit by Category

We calculate the average profit generated by bookings in each hotel category.

```python
profit_by_category = df.groupby("Category")["Profit"].mean()

profit_by_category
```

```python
profit_by_category_sorted = profit_by_category.sort_values(
    ascending=False
)

profit_by_category_sorted
```

```python
top_profit_category = profit_by_category_sorted.index[0]
top_average_profit = profit_by_category_sorted.iloc[0]

print("Category with highest average profit:", top_profit_category)
print("Average profit:", top_average_profit)
```

## 5.1 Monthly Sales — Matplotlib Line Chart

This chart shows how total StayEase sales change month by month.

```python
monthly_sales = (
    df.groupby(df["OrderDate"].dt.to_period("M"))["Sales"]
    .sum()
)

monthly_sales
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    monthly_sales.index.astype(str),
    monthly_sales.values,
    marker="o"
)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Total Sales (INR)")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
