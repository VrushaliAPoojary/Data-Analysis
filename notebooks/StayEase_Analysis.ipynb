## 4.2 Correlation Analysis

Correlation measures the strength and direction of the linear relationship
between numerical variables.

```python
correlation_data = df[
    ["Sales", "Profit", "Quantity", "DeliveryDays"]
]

correlation_matrix = correlation_data.corr()

correlation_matrix
```

```python
corr_pairs = correlation_matrix.where(
    np.triu(np.ones(correlation_matrix.shape), k=1).astype(bool)
)

corr_pairs
```

```python
strongest_pair = (
    corr_pairs.stack()
    .abs()
    .sort_values(ascending=False)
)

strongest_pair
```

## 6. Business Insights

1. StayEase generated total Sales of ₹70.95 million across the dataset, with an average Sales value of ₹35,960.08 per booking. The median Sales value was ₹23,374.49, indicating that higher-value bookings raise the average above the median.

2. The South region recorded the highest total Sales at ₹19.02 million, followed by East at ₹18.91 million, North at ₹17.06 million, and West at ₹15.96 million.

3. Wallet was the most-used payment method, accounting for 411 bookings (20.83% of all bookings). Net Banking followed at 20.53%, UPI at 19.97%, Debit Card at 19.82%, and Credit Card at 18.85%.

4. Banquet/Event generated the highest average Profit at ₹16,449.37 per booking, substantially exceeding the average Profit of the other categories.

5. Sales and Profit showed a strong positive correlation of 0.8596, indicating that higher Sales values are generally associated with higher Profit values.

## 7. Business Recommendations

### Recommendation 1 — Strengthen the South Region

The South region generated the highest total Sales at ₹19.02 million. StayEase should identify the properties, customer segments, and sales strategies contributing to this strong performance and consider replicating successful practices in lower-performing regions.

### Recommendation 2 — Focus on the Banquet/Event Category

The Banquet/Event category generated the highest average Profit at ₹16,449.37 per booking. StayEase should investigate the pricing, demand, and cost structure behind this performance and strengthen profitable event offerings where appropriate.
